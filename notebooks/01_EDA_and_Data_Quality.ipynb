This notebook audits data hygiene, identifies corrupted/mismatched experiment logs, and tests for Sample Ratio Mismatch (SRM).

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_validator import validate_and_clean_dataset

raw_ab = '../data/raw/ab_data.csv'
raw_countries = '../data/raw/countries.csv'
clean_out = '../data/processed/clean_experiment_data.csv'

clean_df, report = validate_and_clean_dataset(raw_ab, raw_countries, clean_out)
print('Total Raw Rows:', report.total_raw_rows)
print('Mismatched Records Purged:', report.mismatched_assignments)
print('Clean Verified Users:', report.clean_rows)
print('SRM Chi-Square Statistic:', report.srm_chi2_stat)
print('SRM P-Value:', report.srm_p_value)
print('SRM Passed:', report.srm_passed)
```

## Visualizing Group Distribution and Conversion Baselines

```python
plt.figure(figsize=(8, 4))
sns.barplot(data=clean_df, x='group', y='converted', ci=None, palette=['#1a73e8', '#ea4335'])
plt.title('Conversion Rate by Experiment Group')
plt.ylabel('Conversion Rate')
plt.ylim(0, 0.15)
for idx, grp in enumerate(['control', 'treatment']):
    rate = clean_df[clean_df['group'] == grp]['converted'].mean()
    plt.text(idx, rate + 0.005, f'{rate:.2%}', ha='center', fontweight='bold')
plt.show()
```
