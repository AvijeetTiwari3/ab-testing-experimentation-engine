# Notebook 02: Frequentist Hypothesis Testing & Power Analysis
Rigorous inference using Two-Proportion Z-Test, Cohen's h Effect Size, and Sizing Curves.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
from src.frequentist_engine import two_proportion_ztest
from src.power_analysis import calculate_sample_size_proportions

df = pd.read_csv('../data/processed/clean_experiment_data.csv')
c_df = df[df['group'] == 'control']
t_df = df[df['group'] == 'treatment']

c_conv, c_n = int(c_df['converted'].sum()), len(c_df)
t_conv, t_n = int(t_df['converted'].sum()), len(t_df)

res = two_proportion_ztest(c_conv, c_n, t_conv, t_n, alpha=0.05, mde=0.005)
for k, v in res.to_dict().items():
    print(f'{k:30s}: {v}')
```
